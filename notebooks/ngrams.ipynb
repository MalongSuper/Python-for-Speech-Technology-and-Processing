### N-Grams Models

N-Grams is a sequential model that predicts the next item (word or character) in a sequence by analyzing the previous items. It operates as a probabilistic model based on the Markov assumption, relying on a fixed context window to calculate the likelihood of a word's occurrence.

In modern text-to-speech systems, N-Grams are used to enhance the accuracy of predictions by considering the context provided by preceding words. The model can be applied to various natural language processing tasks, including text generation, speech recognition, and machine translation.

For more information:
https://www.geeksforgeeks.org/nlp/n-gram-in-nlp/

```python
def generate_ngrams(text, n):
    tokens = text.split()
    ngrams = [tuple(tokens[i:i + n]) for i in range(len(tokens) - n + 1)]
    return ngrams


text = input("Enter a string: ")

unigrams = generate_ngrams(text, 1)
bigrams = generate_ngrams(text, 2)
trigrams = generate_ngrams(text, 3)

print("Unigrams:", unigrams)
print("Bigrams:", bigrams)
print("Trigrams:", trigrams)
```

### Laplace Smoothing for N-grams
When working with N-grams, one of the major challenges is data sparsity especially with higher-order N-grams like 4-grams or 5-grams. As the value of N increases the number of possible N-grams grows exponentially and many of them may not appear in the training data resulting in zero probabilities for unseen sequences.

To resolve this we use Laplace Smoothing also known as Additive Smoothing. It adds a constant (typically 1) to each count ensuring that no N-gram has a zero probability even if it was not seen in the training set.

```python
from collections import Counter


def laplace_smoothing(text, vocab_size, n=1):
    ngrams = generate_ngrams(text, n)
    ngram_counts = Counter(ngrams)
    smoothed_ngrams = {ngram: (count + 1) / (len(ngrams) + vocab_size)
                       for ngram, count in ngram_counts.items()}
    return smoothed_ngrams


# Main function
smoothed_ngrams1 = laplace_smoothing(text, vocab_size=5, n=1)
print("Smoothed N-grams (Unigrams):", smoothed_ngrams1)
smoothed_ngrams2 = laplace_smoothing(text, vocab_size=5, n=2)
print("Smoothed N-grams (Bigrams):", smoothed_ngrams2)
smoothed_ngrams3 = laplace_smoothing(text, vocab_size=5, n=3)
print("Smoothed N-grams (Trigrams):", smoothed_ngrams3)
```

```python
# Turn the result to pandas for better visualization
import pandas as pd

data1 = smoothed_ngrams1
df = pd.DataFrame(list(data1.items()), columns=['Unigrams', 'Probability_Unigrams'])
# Add smoothed_ngrams2 and smoothed_ngrams3 with its probability as columns
data2 = smoothed_ngrams2
df2 = pd.DataFrame(list(data2.items()), columns=['Bigrams', 'Probability_Bigrams'])
data3 = smoothed_ngrams3
df3 = pd.DataFrame(list(data3.items()), columns=['Trigrams', 'Probability_Trigrams'])
# Concatenate them with df
df = pd.concat([df, df2, df3], axis=1)
df
```
