### Recurrent Neural Networks (RNNs)
Recurrent Neural Networks (RNNs) are a class of neural networks designed to handle sequential data. They are particularly effective for tasks such as language modeling, time series prediction, and speech recognition. RNNs have a unique architecture that allows them to maintain a hidden state that captures information from previous time steps, making them well-suited for processing sequences of data.

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
sentence = ["I am currently studying Artificial Intelligence in Lincoln University"]
```

Tokenization converts each word into an integer.

```python
tokenizer = Tokenizer(num_words=10000)
tokenizer.fit_on_texts(sentence)

sequences = tokenizer.texts_to_sequences(sentence)

print("Word Index:", tokenizer.word_index)
print("Sequence:", sequences)
```

RNNs require inputs of equal length.

```python
max_length = 10
X = pad_sequences(sequences, maxlen=max_length)

print("Padded Sequence:")
print(X)
```

Since this is just a demonstration, we assign a simple label.
- Example: 1 = sentence related to AI study

```python
y = np.array([1])
```

Build the RNN Model

```python
model = Sequential([Embedding(input_dim=10000, output_dim=32), SimpleRNN(units=32), 
                    Dense(units=1, activation='sigmoid')])
```

Explanation:
1. **Embedding Layer**: This layer converts the integer-encoded words into dense vectors of fixed size. It helps the model learn relationships between words based on their context in the training data.
2. **LSTM Layer**: Long Short-Term Memory (LSTM) is a type of RNN that can capture long-term dependencies in the data. It helps the model remember information from previous time steps, which is crucial for understanding the context in sequences.
3. **Dense Layer**: This is the output layer that produces the final prediction. In this case, it uses a sigmoid activation function to output a value between 0 and 1, which can be interpreted as the probability of the sentence being related to AI study.

Compile the Model

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

Train the model

```python
model.fit(X, y, epochs=10)
```
